## Problem 2. Electrical circuits: linear and non-linear systems

Consider a DC circuit in which two nodes, $A$ and $B$, are connected by three parallel branches:

- **Branch 1:** a battery of electromotive force $\varepsilon_1$ in series with a resistor $R_1$, carrying a current $I_1$ from $A$ to $B$.
- **Branch 2:** a battery of electromotive force $\varepsilon_2$ in series with a resistor $R_2$, carrying a current $I_2$ from $A$ to $B$.
- **Branch 3:** a resistor $R_3$, carrying a current $I_3$ from $B$ to $A$.

Kirchhoff's junction rule at node $A$ and Kirchhoff's loop rule applied to the two independent loops give a **system of linear equations** for the three unknown currents.

For this problem, use the following values:

| Quantity | Value |
|----------|-------|
| $\varepsilon_1$ | $5\,\rm V$ |
| $\varepsilon_2$ | $3\,\rm V$ |
| $R_1$ | $1.0\,\rm k\Omega$ |
| $R_2$ | $2.2\,\rm k\Omega$ |
| $R_3$ | $3.3\,\rm k\Omega$ |

Now, suppose that a **semiconductor diode** is placed in series with $R_3$ in branch 3. The voltage drop across a diode is not proportional to the current through it, but follows the Shockley equation:

$$I_3 = I_s\left[\exp\left(\frac{V_{\rm D}}{n\,V_{\rm T}}\right) - 1\right]
\qquad\Longleftrightarrow\qquad
V_{\rm D} = n\,V_{\rm T}\,\ln\left(\frac{I_3}{I_s} + 1\right)$$

where $I_s = 10^{-12}\,\rm A$ is the reverse saturation current, $n = 1$ is the ideality factor, and $V_{\rm T} = 0.02585\,\rm V$ is the thermal voltage at $T = 300\,\rm K$. Adding this element turns the circuit equations into a **system of non-linear equations**.
