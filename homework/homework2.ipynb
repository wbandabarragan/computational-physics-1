## Problem statement: cloud-cloud collision

Molecular clouds in the interstellar medium move with respect to one another, so they sometimes collide. These collisions drive shocks into both clouds, compress the gas, and can trigger star formation. Signatures of such collisions have been reported in the Galactic Centre:

https://scixplorer.org/abs/2020MNRAS.499.4918A

Here, we numerically study one of such collisions, where a small cloud travels across a 2D computational domain (along the X-axis from left to right) and impacts a larger cloud that is initially at rest. In the simulation, the gas of each cloud is tracked with its own pigment (tracer): `tr1` for the small cloud and `tr2` for the large one.

The flow is adiabatic (the polytropic index is $\gamma=\frac{5}{3}$), the gas is not magnetised, and both clouds start in pressure equilibrium with the ambient medium, with a density contrast of $\chi=10$ for the small cloud and $\chi=5$ for the large one.
